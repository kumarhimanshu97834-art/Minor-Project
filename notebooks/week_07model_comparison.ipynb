```python

# WEEK 7 - STEP 5
# Evaluate original Week 3 Custom CNN on current test split

from pathlib import Path
import pandas as pd
import numpy as np
import torch
import torch.nn as nn

from PIL import Image
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

ROOT = Path.cwd().parent
RESULTS = ROOT / "results"
MODEL_PATH = ROOT / "models" / "week3_custom_cnn.pth"

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Load current metadata
df = pd.read_csv(RESULTS / "final_face_metadata.csv")
test_df = df[df["split"].str.lower() == "test"].copy()

test_df["full_path"] = test_df["face_path"].apply(
    lambda p: str(ROOT / p)
)

print("Device:", device)
print("Test images:", len(test_df))

# Original Week 3 preprocessing
transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor()
])

class FaceTestDataset(Dataset):
    def __init__(self, data):
        self.data = data.reset_index(drop=True)

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        row = self.data.iloc[idx]

        with Image.open(row["full_path"]) as im:
            image = transform(im.convert("RGB"))

        label = torch.tensor(
            float(row["label"]),
            dtype=torch.float32
        )

        return image, label

test_loader = DataLoader(
    FaceTestDataset(test_df),
    batch_size=32,
    shuffle=False,
    num_workers=0
)

# Original Week 3 architecture
class CustomCNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.AdaptiveAvgPool2d((1, 1)),
            nn.Flatten()
        )

        self.classifier = nn.Linear(128, 1)

    def forward(self, x):
        return self.classifier(
            self.features(x)
        ).squeeze(1)

model = CustomCNN().to(device)

# Load the original Week 3 checkpoint
state_dict = torch.load(
    MODEL_PATH,
    map_location="cpu",
    weights_only=True
)

model.load_state_dict(state_dict, strict=True)
model.eval()

print("Original Custom CNN checkpoint loaded!")

# Evaluate
y_true = []
y_pred = []
y_prob = []

with torch.inference_mode():
    for images, labels in test_loader:
        images = images.to(device)

        logits = model(images)
        probabilities = torch.sigmoid(logits)

        predictions = (probabilities >= 0.5).long()

        y_true.extend(labels.numpy().astype(int))
        y_pred.extend(predictions.cpu().numpy())
        y_prob.extend(probabilities.cpu().numpy())

y_true = np.array(y_true)
y_pred = np.array(y_pred)
y_prob = np.array(y_prob)

# Metrics
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(
    y_true, y_pred, pos_label=1, zero_division=0
)
recall = recall_score(
    y_true, y_pred, pos_label=1, zero_division=0
)
f1 = f1_score(
    y_true, y_pred, pos_label=1, zero_division=0
)
roc_auc = roc_auc_score(y_true, y_prob)

print("\n========== ORIGINAL CUSTOM CNN RESULTS ==========")
print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1-score : {f1:.4f}")
print(f"ROC-AUC  : {roc_auc:.4f}")

cm = confusion_matrix(y_true, y_pred)

print("\nConfusion Matrix:")
print(cm)

print("\nClassification Report:")
print(classification_report(
    y_true,
    y_pred,
    labels=[0, 1],
    target_names=["Real", "Fake"],
    zero_division=0
))

# Save results
pd.DataFrame([{
    "Model": "Original Custom CNN",
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "F1-Score": f1,
    "ROC-AUC": roc_auc,
    "Test_Images": len(y_true)
}]).to_csv(
    RESULTS / "week7_custom_cnn_final_metrics.csv",
    index=False
)

pd.DataFrame(
    cm,
    index=["Actual Real", "Actual Fake"],
    columns=["Predicted Real", "Predicted Fake"]
).to_csv(
    RESULTS / "week7_custom_cnn_confusion_matrix.csv"
)

pd.DataFrame({
    "true_label": y_true,
    "predicted_label": y_pred,
    "probability_fake": y_prob
}).to_csv(
    RESULTS / "week7_custom_cnn_predictions.csv",
    index=False
)

print("\nAll Custom CNN evaluation files saved successfully.")
```

```python

from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent
RESULTS = ROOT / "results"
FIGURES = RESULTS / "figures"
FIGURES.mkdir(exist_ok=True)

# Load the two verified evaluation results
cnn = pd.read_csv(
    RESULTS / "week7_custom_cnn_final_metrics.csv"
).iloc[0]

resnet = pd.read_csv(
    RESULTS / "week6_final_metrics.csv"
).iloc[0]

comparison = pd.DataFrame([
    {
        "Model": "Custom CNN",
        "Accuracy": cnn["Accuracy"],
        "Precision": cnn["Precision"],
        "Recall": cnn["Recall"],
        "F1-Score": cnn["F1-Score"],
        "ROC-AUC": cnn["ROC-AUC"]
    },
    {
        "Model": "ResNet50",
        "Accuracy": resnet["Accuracy"],
        "Precision": resnet["Precision"],
        "Recall": resnet["Recall"],
        "F1-Score": resnet["F1-Score"],
        "ROC-AUC": resnet["ROC-AUC"]
    }
])

comparison.to_csv(
    RESULTS / "week7_final_model_comparison.csv",
    index=False
)

print("FINAL MODEL COMPARISON")
print(comparison.to_string(index=False))

# Comparison chart
metrics = ["Accuracy", "Precision", "Recall", "F1-Score", "ROC-AUC"]

comparison.set_index("Model")[metrics].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Custom CNN vs ResNet50 - Final Comparison")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)
plt.legend(loc="lower right")
plt.tight_layout()

plt.savefig(
    FIGURES / "week7_final_model_comparison.png",
    dpi=300
)

plt.show()

print("\nWeek 7 comparison files saved successfully.")
```
